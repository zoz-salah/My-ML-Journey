## HOW TO DETECT OUTLIERS

```python
Q1 = dc['quantity'].quantile(0.25)
Q3 = dc['quantity'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outlierss = dc[(dc['quantity'] < lower_bound) | (dc['quantity'] > upper_bound)]
print(outlierss)
```

```python
Q11 = dc['profit'].quantile(0.25)
Q33 = dc['profit'].quantile(0.75)
IQR = Q33 - Q11
lower_bound = Q11 - 1.5 * IQR
upper_bound = Q33 + 1.5 * IQR

outliers = dc[(dc['profit'] < lower_bound) | (dc['profit'] > upper_bound)]
print(outliers)
```

```python
fig = px.histogram(outliers, x='profit', title="Histogram of Profit")
fig.show()
```

```python
fig = px.histogram(outlierss, x='quantity', title="Histogram of Quantity")
fig.show()
```

```python
from scipy import stats
dc['z_score'] = stats.zscore(dc['profit'])
```

```python
print(dc[['profit', 'z_score']].head(10))
```

```python
figg = px.histogram(dc[['profit', 'z_score']], x='profit', title="Histogram of Profit")
figg.show()
```

```python
z = (dc['profit'] - dc['profit'].mean()) / dc['profit'].std()
dc['z_outlier'] = z.abs() > 3
```

```python
figgg = px.histogram(z, x='profit', title="Histogram of Profit")
figgg.show()
```
