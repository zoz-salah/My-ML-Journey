## DATA VISUALIZATION  using plotly

```python
import pandas as pd
dc = pd.read_csv('D:/TECHTRECK/retail_visualization_dataset.csv')
pd.set_option('display.max_columns', 50)  
```

```python
## groupby 

total_revenue_by_region = dc.groupby('city')['revenue'].sum().sort_values(ascending=False)
print(total_revenue_by_region)
```

```python
import plotly.express as px
```

```python
fig =px.bar(total_revenue_by_region,x=total_revenue_by_region.index,y=total_revenue_by_region.values, color_discrete_sequence=['red'], title='Total Revenue by Region')
fig.update_layout(xaxis_title='Region', yaxis_title='Total Revenue')
```

```python
total_revenue_by_region.plot(kind='bar')
```

```python
channel_pay = dc['channel'].value_counts()
channel_pay
```

```python
channel_pay.plot(kind='bar', title='Channel Distribution')
```

```python
px.bar(channel_pay, x=channel_pay.index, y=channel_pay.values, color_discrete_sequence=['red'], title='Channel Distribution')
```

```python
import plotly.express as px

fig = px.bar(
    channel_pay,
    x=channel_pay.index,
    y=channel_pay.values,
    color_discrete_sequence=['red'], 
    title='Channel Distribution'
)

fig.update_layout(
    template='plotly_white',        
    title={
        'x': 0.5,                   
        'font': {'size': 24}
    },
    xaxis_title='Channel',
    yaxis_title='Count',
    font=dict(
        family='Arial',
        size=14,
        color='black'
    ),
    bargap=0.25,
    plot_bgcolor='white',
    paper_bgcolor='white',
    xaxis=dict(
        showgrid=False,
        linecolor='black',
        mirror=True
    ),
    yaxis=dict(
        showgrid=True,
        gridcolor='lightgray',
        linecolor='black',
        mirror=True
    ),
    width=900,
    height=500
)

fig.update_traces(
    marker_line_color='black',
    marker_line_width=1.2,
    opacity=0.9,
    hovertemplate='<b>%{x}</b><br>Count: %{y}<extra></extra>'
)

fig.show()
```

```python
customer_segment_sales = dc.groupby('customer_segment')['gross_sales'].mean().sort_values(ascending=False)
px.bar(customer_segment_sales, x=customer_segment_sales.index, y=customer_segment_sales.values, title='Average Gross Sales & Customer Segment',color_discrete_sequence=['red'] ,labels={'x':'Customer Segment', 'y':'Average Gross Sales'})
```

```python
px.area(customer_segment_sales, x=customer_segment_sales.index, y=customer_segment_sales.values, title='Average Gross Sales by Customer Segment', labels={'x':'Customer Segment', 'y':'Average Gross Sales'}, color_discrete_sequence=['red'], template='plotly_white')
```

```python
px.histogram(customer_segment_sales, x=customer_segment_sales.index, y=customer_segment_sales.values, title='Average Gross Sales by Customer Segment', labels={'x':'Customer Segment', 'y':'Average Gross Sales'}, color_discrete_sequence=['red'], template='plotly_white' )
```

```python
px.pie(customer_segment_sales, values=customer_segment_sales.values, names=customer_segment_sales.index, title='Average Gross Sales by Customer Segment', color_discrete_sequence=px.colors.sequential.RdBu, template='plotly_white')
```
